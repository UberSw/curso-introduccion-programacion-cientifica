## **Ejercicio 1 - Inventario de la tienda**




```python
nombre_producto = "Cuaderno"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print("TIPOS DE VARIABLES\n")
print(f"nombre_producto:    {type(nombre_producto)}")
print(f"precio_cuaderno:    {type(precio_cuaderno)}")
print(f"precio_lapiz:       {type(precio_lapiz)}")
print(f"precio_cafe:        {type(precio_cafe)}")
print(f"hay_descuento:      {type(hay_descuento)}")
print(f"cantidad_cuadernos: {type(cantidad_cuadernos)}")
print(f"cantidad_lapices:   {type(cantidad_lapices)}")
```

## **Ejercicio 2 - Total de la compra y descuento**


```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
print(f"Total cuadernos:     {total_cuadernos:,.2f}")

total_lapices = precio_lapiz * cantidad_lapices
print(f"Total lápices:        {total_lapices:,.2f}")

total_general = total_cuadernos + total_lapices + precio_cafe
print(f"Total general:       {total_general:,.2f}")

descuento = total_general * 0.1
print(f"Descuento:            {descuento:,.2f}")

total_con_descuento = total_general - descuento
print(f"Total con descuento: {total_con_descuento:,.2f}")
```

## **Ejercicio 3 - Reparto en paquetes**


```python
paquetes_cuadernos = cantidad_cuadernos // 4
cuadernos_sueltos = cantidad_cuadernos % 4
precio_cafe_cuadrado = precio_cafe ** 2

print(f"Paquetes de 4 cuadernos:     {paquetes_cuadernos}")
print(f"Total cuadernos sueltos:     {cuadernos_sueltos}")
print(f"Precio del café al cuadrado: {precio_cafe_cuadrado:,.2f}")
```

## **Ejercicio 5 - Promedio y comparación de temperaturas**


```python
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

temp_prom_tres_dias = (temp_lunes + temp_martes + temp_miercoles) / 3
print(f"Temperatura promedio de los tres días: {temp_prom_tres_dias:.2f}")

diferencia = temp_martes - temp_lunes

martes_mas_caluroso = temp_martes > temp_promedio_semana
print(f"Martes más caluroso que el promedio:   {martes_mas_caluroso}")

miercoles_igual_promedio = temp_miercoles == temp_promedio_semana
print(f"Miércoles igual al promedio:           {miercoles_igual_promedio}")

lunes_dia_mas_frio = (temp_lunes < temp_martes) and (temp_lunes < temp_miercoles)
print(f"Lunes día más frío de los tres:        {lunes_dia_mas_frio}")
```

## **Ejercicio 6 - Temperatura templada y día caluroso**


```python
temp_lunes_templada = (temp_lunes >= 15) and (temp_lunes <= 25)
print(f"Lunes tuvo una temperatura templada: {temp_lunes_templada}")

temp_mayor_20grados = (temp_lunes > 20) or (temp_martes > 20) or (temp_miercoles > 20)
print(f"Algun día superó los 20 grados:      {temp_mayor_20grados}")
```

## **Ejercicio 7 - Factura de servicios y consumo normal**

## **Ejercicio 10 - Intercambiar dos precios**


```python
precio_tornillo = 250
precio_tuerca = 180

print("Precios antes de intercambiar")
print(f"Precio de tornillo: {precio_tornillo}")
print(f"Precio de tuerca:   {precio_tuerca}")

temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

print("\nPrecios después de intercambiar")
print(f"Precio de tornillo: {precio_tornillo}")
print(f"Precio de tuerca:   {precio_tuerca}")


```
