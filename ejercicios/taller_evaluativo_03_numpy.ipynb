## **Ejercicio 1 — Crear los arreglos y leer su etiqueta**
Se importa la librería NumPy y se construye la matriz bidimensional **lecturas_crudas** a partir de una lista de listas con las mediciones semanales de los cinco sensores (SBC-A a SBC-E). Se lee la etiqueta de la hoja (arreglo lecturas_crudas) extrayendo los atributos: forma (shape), número de dimensiones (ndim), total de datos (size) y tipo de dato (dtype). Luego, se crea el vector de días utilizando **np.arange**  y los arreglos 1D de **pendientes** e **interceptos** con los parámetros de calibración de cada sensor. Se finaliza mostrando todos los resultados mediante *print().*

```python
import numpy as np

lecturas_crudas = np.array([
    [10.0, 14.5, 22.0, 30.5, 28.0, 18.0,  9.5],  # SBC-A
    [12.0, 20.0, 35.0, 48.0, 41.0, 25.0, 14.0],  # SBC-B
    [ 8.0, 11.0, 19.0, 27.0, 33.0, 21.0,  6.0],  # SBC-C
    [15.0, 24.0, 31.0, 44.0, 52.0, 36.0, 17.0],  # SBC-D
    [ 5.0,  9.0, 16.0, 22.0, 45.0, 30.0,  4.0]   # SBC-E
])

forma = lecturas_crudas.shape
dimensiones = lecturas_crudas.ndim
total = lecturas_crudas.size
tipo = lecturas_crudas.dtype

dias = np.arange(1, 8)

pendientes = np.array([1.2, 0.9, 1.1, 1.0, 1.3])
interceptos = np.array([-0.5, 1.0, -2.0, 0.5, -1.5])

print("*****************************************")
print("*               INFORME                 *")
print("*****************************************\n")
print(f"Forma:        {forma}")
print(f"Dimensiones:  {dimensiones}")
print(f"Total celdas: {total}")
print(f"Tipo:         {tipo}")
print(f"Días:         {dias}")
print(f"Pendientes:   {pendientes}")
print(f"Interceptos:  {interceptos}")

```

## **Ejercicio 2 — Señalar celdas y recortar bloques**
En el ejercicio se aplican técnicas de indexación y *slicing* sobre el arreglo bidimensional **lecturas_crudas** sin utilizar bucles. Se accede a posiciones puntuales mediante índices directos e índices negativos, se extraen arreglos unidimensionales completos correspondientes a una columna específica (día 5) y a una fila completa (sensor SBC-B), y finalmente se recorta una submatriz bidimensional de forma $(3, 4)$ que abarca los sensores SBC-B a SBC-D durante los días 3 al 6. Se muestra con *print()* los resultados y las dimensiones del bloque.

```python
lectura_c_dia4 = lecturas_crudas[2, 3]

ultimo_primer = lecturas_crudas[-1, 0]

dia5 = lecturas_crudas[:, 4]

semana_b = lecturas_crudas[1, :]

bloque = lecturas_crudas[1:4, 2:6]

print("**********************************************")
print("*                    INFORME                 *")
print("**********************************************\n")
print(f"lectura_c_dia4:  {lectura_c_dia4}")
print(f"ultimo_primer:   {ultimo_primer}")
print(f"dia5:            {dia5}")
print(f"semana_b:        {semana_b}")
print()
print(f"bloque:\n{bloque}")
print()
print(f"forma de bloque: {bloque.shape}")
```

## **Ejercicio 3 — Limpiar una campaña con máscaras booleanas**
En este ejercicio se procesan las lecturas de campaña del sensor SBC-A representándolas como un arreglo unidimensional de NumPy (campana_sbc_a). Mediante comparaciones vectorizadas y el operador bit a bit &, se generan máscaras booleanas para identificar datos faltantes (faltantes) y lecturas físicamente útiles dentro del intervalo [0.0, 500.0] (es_util), descartando simultáneamente el valor centinela (-999.0) y el ruido excesivo (600.0). Utilizando la máscara resultante como índice de filtrado, se extraen las lecturas válidas (lecturas_utiles), se contabilizan los datos faltantes a través de la suma de booleanos y se calcula el promedio aritmético con el método mean(), mostrando los resultados con *print().*

```python
campana_sbc_a = np.array([10.0, -999.0, 25.0, 40.0, -999.0, 8.0, 52.0, 600.0])

faltantes = (campana_sbc_a == -999.0)

es_util = (campana_sbc_a >= 0.0) & (campana_sbc_a <= 500.0)

lecturas_utiles = campana_sbc_a[es_util]

promedio_utiles = lecturas_utiles.mean()

print("******************************************************")
print("*                        INFORME                     *")
print("******************************************************\n")
print(f"Cantidad de lecturas faltantes: {faltantes.sum()}")
print(f"Lecturas útiles:                {lecturas_utiles}")
print(f"Promedio de lecturas útiles:    {promedio_utiles}")
```

## **Ejercicio 4 — Calibrar todo con operaciones vectorizadas y broadcasting**

La operación directa lecturas_crudas * pendientes produce un error de tipo ValueError debido a las reglas de *broadcasting* de NumPy. Al alinear las formas de derecha a izquierda, NumPy compara el último eje de lecturas_crudas (7 columnas) con el único eje de pendientes (5 elementos). Como los tamaños no coinciden (7 frente a 5) y ninguno es igual a 1, no es posible estirar los arreglos.

Para solucionar esta incompatibilidad, se transforman los vectores 1D de pendientes e interceptos en arreglos columna bidimensionales con forma (5, 1) mediante reshape(5, 1) (o np.newaxis). De este modo, la dimensión 1 se expande a lo largo de las 7 columnas y las 5 filas coinciden exactamente con cada uno de los sensores, aplicando la calibración a toda la matriz en una única operación vectorizada sin necesidad de bucles. Finalmente, se calibran las lecturas útiles obtenidas en el Ejercicio 3 para el sensor SBC-A y se verifica su promedio.

```python
pendientes_col = pendientes.reshape(5, 1)
interceptos_col = interceptos.reshape(5, 1)

calibradas = lecturas_crudas * pendientes_col + interceptos_col

campana_calibrada = lecturas_utiles * 1.2 + (-0.5)

promedio_calibrado_sbc_a = campana_calibrada.mean()

print("****************************************************************")
print("*                             INFORME                          *")
print("****************************************************************\n")
print(f"Forma de calibradas:        {calibradas.shape}")
print(f"Fila calibradas[0] (SBC-A): {np.round(calibradas[0], 2)}")
print(f"campana_calibrada (SBC-A):  {campana_calibrada}")
print(f"promedio_calibrado_sbc_a:   {promedio_calibrado_sbc_a:.1f}")
```

## **Ejercicio 5 — Resumir la semana con estadística por eje**
En el ejercicio se sintetiza la información contenida en la matriz **calibradas** aplicando métodos de agregación y reducción sobre diferentes ejes de NumPy[cite: 26]. Se obtiene primero el promedio global sobre la totalidad de las 35 celdas mediante mean(). Luego, se calcula el promedio y el valor máximo por sensor reduciendo a lo largo de las columnas (axis=1), obteniendo un valor por cada una de las 5 filas. Paralelamente, se calcula el promedio por día reduciendo verticalmente a lo largo de las filas (axis=0), lo que genera un valor representativo para cada una de las 7 columnas. Por último, se aplica una máscara booleana para filtrar y computar la media únicamente de aquellas concentraciones calibradas que sobrepasan el umbral de 35.4 ug/m3, presentando todas las salidas redondeadas a dos decimales con **np.round**.

```python
promedio_general = calibradas.mean()

promedio_por_sensor = calibradas.mean(axis=1)

promedio_por_dia = calibradas.mean(axis=0)

maximo_por_sensor = calibradas.max(axis=1)

media_sobre_limite = calibradas[calibradas > 35.4].mean()

print("****************************************************************")
print("*                             INFORME                          *")
print("****************************************************************\n")
print(f"Promedio general:    {np.round(promedio_general, 2)}")
print(f"Promedio por sensor: {np.round(promedio_por_sensor, 2)}")
print(f"Promedio por día:    {np.round(promedio_por_dia, 2)}")
print(f"Máximo por sensor:   {np.round(maximo_por_sensor, 2)}")
print(f"Media sobre límite:  {np.round(media_sobre_limite, 2)}")
```

## **Ejercicio 6 — Funciones universales y reemplazo con máscara**
En este ejercicio se aplican funciones universales (*ufuncs*) de NumPy para evaluar cuánto sobrepasan las concentraciones respecto a la frontera de 35.4 ug/m3 (límite entre categoría Moderada y Dañina para grupos sensibles). Se calcula el arreglo **excedente** utilizando np.maximum(calibradas - 35.4, 0), conservando el exceso positivo o fijando en 0 si la medición no supera dicho límite, y luego se suma horizontalmente por sensor (axis=1). Asimismo, con np.abs() se determina la distancia absoluta al límite para el sensor SBC-B (desvio_sensor_b). Finalmente, se crea una copia del arreglo original mediante copy() para construir la matriz en_alerta, reemplazando mediante una máscara booleana por 0.0 todas aquellas celdas que no superan 35.4, y se contabilizan las celdas en alerta usando sum(). Se finaliza usando *print()* para mostrar los resultados.

```python
excedente = np.maximum(calibradas - 35.4, 0)

excedente_por_sensor = np.round(excedente.sum(axis=1), 1)

desvio_sensor_b = np.round(np.abs(calibradas[1] - 35.4), 1)

en_alerta = calibradas.copy()
en_alerta[en_alerta <= 35.4] = 0.0

total_en_alerta = (en_alerta > 0).sum()

print("************************************************************************")
print("*                                 INFORME                              *")
print("************************************************************************\n")
print(f"Excedente por sensor:               {excedente_por_sensor}")
print(f"Desvío absoluto sensor B al límite: {desvio_sensor_b}")
print(f"Total de celdas en alerta (> 0):    {total_en_alerta}")
```

## **Ejercicio 7 — Integración: el informe de la red**
En este último ejercicio final se clasifican las 35 lecturas del arreglo **calibradas** según los rangos de la escala oficial de calidad del aire para PM2.5 (Buena: <= 12.0, Moderada: > 12.0 y <= 35.4, Dañina para grupos sensibles: > 35.4 y <= 55.4, Dañina para la salud: > 55.4), asegurando que cada medición pertenezca exactamente a una sola categoría y contabilizándolas con sum(). Adicionalmente, se identifica el día más contaminado filtrando el arreglo **dias** mediante una máscara booleana que detecta el valor máximo en **promedio_por_dia,** extrayendo el día pico y su promedio respectivo. Para terminar, se imprimen los resultados en pantalla con el formato solicitado.

```python
buenas = (calibradas <= 12.0).sum()
moderadas = ((calibradas > 12.0) & (calibradas <= 35.4)).sum()
daninas_sensibles = ((calibradas > 35.4) & (calibradas <= 55.4)).sum()
daninas_salud = (calibradas > 55.4).sum()

mascara_pico = (promedio_por_dia == promedio_por_dia.max())
dia_pico = dias[mascara_pico][0]
promedio_pico = np.round(promedio_por_dia.max(), 1)


print("****************************************************************")
print("*                             INFORME                          *")
print("****************************************************************\n")
print(f"LECTURAS POR CATEGORÍA\n")
print(f"Buena:                        {buenas}")
print(f"Moderada:                     {moderadas}")
print(f"Dañina para grupos sensibles: {daninas_sensibles}")
print(f"Dañina para la salud:         {daninas_salud}")
print()
print(f"Día más contaminado:          {dia_pico} con {promedio_pico} ug/m3")
```
