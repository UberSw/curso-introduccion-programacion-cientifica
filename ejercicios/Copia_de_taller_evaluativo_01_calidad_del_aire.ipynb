## **Ejercicio 2 — Conversión de unidades y error frente al equipo patrón**

En este ejercicio se realiza la conversión de unidades entre el sensor de campo y el equipo patrón de referencia. Se aplica el factor de conversión obtenido en el Ejercicio 1 para transformar la medición de mg/m3 a ug/m3. A partir del valor estandarizado, se evalúa la exactitud del sensor calculando tres métricas de discrepancia mediante operadores aritméticos básicos (*, -, /, **): el error absoluto, el error relativo porcentual y el error cuadrático.

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion

error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print("CONVERSIÓN DE UNIDADES Y ERRORES\n")
print(f"Lectura del sensor en ug/m3: {lectura_sensor_ug:.2f}")
print(f"Error absoluto:              {error_absoluto:.2f}")
print(f"Error relativo porcentual:   {error_relativo_porcentual:.2f}")
print(f"Error cuadrático:            {error_cuadratico:.2f}")
```

## **Ejercicio 3 - Cobertura temporal del lote descargado**

En este ejercicio se determina el tiempo total de cobertura del lote de datos descargados a partir de la tasa de muestreo de la estación (12 registros por hora). Mediante el uso exclusivo de operadores de división entera (//) y residuo o módulo (%), se descompone la cantidad discreta de mediciones en horas completas y lecturas remanentes, y posteriormente estas horas en días completos y horas sobrantes, evitando aproximaciones de punto flotante.

```python
lecturas_por_hora = 12
horas_completas = numero_lecturas // lecturas_por_hora
lecturas_sobrantes = numero_lecturas % lecturas_por_hora
dias_completos = horas_completas // 24
horas_sobrantes = horas_completas % 24

print("RESULTADOS\n")
print(f"Horas completas:    {horas_completas}")
print(f"Lecturas sobrantes: {lecturas_sobrantes}")
print(f"Días completos:     {dias_completos}")
print(f"Horas sobrantes:    {horas_sobrantes}")

```

## **Ejercicio 4 - Validar una lectura sin decidir todavía**

En este ejercicio se implementa la validación lógica para verificar si una medición individual de concentración es apta para su análisis. Mediante operadores de comparación (==, >=, <=) y operadores lógicos (and, not), se evalúa si la lectura corresponde al valor centinela de dato faltante (-999.0) y si pertenece al rango físicamente medible por el sensor (500.0 ug/m3). La validación se resuelve exclusivamente mediante evaluación booleana, sin recurrir a estructuras condicionales if.

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor
lectura_valida = esta_en_rango and not es_dato_faltante

print("VALIDACIÓN DE LECTURA\n")
print(f"Es dato faltante: {es_dato_faltante}")
print(f"Está en rango:    {esta_en_rango}")
print(f"Lectura válida:   {lectura_valida}")
```

## **Ejercicio 6 -Limpiar la serie horaria**

En esde doce registros consecutivos para depurar las fallas instrumentales reportadas con el código centinela -999.0. Utilizando un ciclo for y la instrucción de control de flujo continue, se filtran las lecturas inválidas registrando la cantidad de descartes sin interrumpir la iteración. Las mediciones válidas se almacenan dinámicamente en una nueva lista mediante .append() y se acumulan para computar, al finalizar el bucle, la concentración promedio representativa del período.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for lectura in lecturas_pm25:
  if lectura == -999.0:
    lecturas_descartadas += 1
    continue
  lecturas_validas.append(lectura)
  suma_pm25 += lectura

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print(f"LIMPIEZA DE SERIE\n")
print(f"Cantidad de lecturas válidas:     {len(lecturas_validas)}")
print(f"Cantidad de lecturas descartadas: {lecturas_descartadas}")
print(f"Promedio de lecturas válidas:     {promedio_pm25:.2f}")


```

## **Ejercicio 7 - Estadística descriptiva de la serie limpia**

En este ejercicio se calcula la estadística descriptiva básica sobre la serie de datos previamente depurada (lecturas_validas). Respetando la restricción de programar de forma nativa sin librerías externas ni funciones integradas de agregación (max(), min(), sum()), se efectúan dos pasadas: la primera para obtener el valor máximo, el mínimo y el rango mediante comparaciones sucesivas; y la segunda para calcular la varianza poblacional y la desviación estándar ($\sigma$) elevando al cuadrado las discrepancias frente al promedio y aplicando una raíz cuadrada por medio del operador de exponenciación ** 0.5.

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for lectura in lecturas_validas:
  if lectura > maximo_pm25:
    maximo_pm25 = lectura
  if lectura < minimo_pm25:
    minimo_pm25 = lectura

rango = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0

for lectura in lecturas_validas:
  suma_desviaciones += (lectura - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print(f"ESTADÍSTICA DESCRIPTIVA DE LA SERIE LIMPIA\n")
print(f"Máximo:              {maximo_pm25}")
print(f"Mínimo:              {minimo_pm25}")
print(f"Rango:               {rango}")
print(f"Desviación estandar: {desviacion_estandar:.2f}")
```

## **Ejercicio 9 - Registros de varias estaciones**

En este ejercicio se integran las estructuras de datos compuestas en Python (list, dict, tuple) para procesar el inventario de la red de monitoreo. Se itera sobre una lista de diccionarios accediendo a sus atributos mediante claves semánticas. Dentro del recorrido principal, se realiza el desempaquetado de la tupla de coordenadas geográficas, se calcula el promedio de concentración de cada estación mediante un ciclo anidado con acumulador nativo y se clasifica según la escala oficial. Para gestionar datos heterogéneos y atributos opcionales sin incurrir en excepciones KeyError, se emplea el método seguro .get() con un valor predeterminado.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
  latitud, longitud = estacion["coordenadas"]
  suma_lecturas = 0.0

  for lectura in estacion["lecturas"]:
    suma_lecturas += lectura

  promedio = suma_lecturas / len(estacion["lecturas"])
  humedad = estacion.get("humedad_relativa", "no disponible")

  if promedio <= 12:
    categoria = "Buena"
  elif promedio > 12.0 and promedio <= 35.4:
    categoria = "Moderada"
  elif promedio > 35.4 and promedio <= 55.4:
    categoria = "Dañina para grupos sensibles"
  else:
    categoria = "Dañina para la salud"

  print(f"Código:    {estacion['codigo']}")
  print(f"Sector:    {estacion['sector']}")
  print(f"Latitud:   {latitud}")
  print(f"Longitud:  {longitud}")
  print(f"Promedio:  {promedio:.2f}")
  print(f"Categoría: {categoria}")
  print(f"Humedad:   {humedad}")
  print()
```
